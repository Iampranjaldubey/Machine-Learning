```python
import numpy as np
```

```python
import pandas as pd
```

```python
df = pd.read_csv('food_vendor_consultation.csv')
```

```python
df.describe()
```

```python
df.sample(5)
```

```python
df.head()
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size= 0.20, random_state= 11)
```

```python
scaler = StandardScaler()

x_train_std = scaler.fit_transform(x_train)
x_test_std = scaler.fit_transform(x_test)
```

```python
from tensorflow import keras
from keras import layers
```

```python
model = keras.Sequential([layers.Dense(units=3, activation='softmax', input_shape=(9,))])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history = model.fit(x_train_std, y_train, epochs=100, batch_size=10)
```

## Graph

```python
l = history.history['loss']
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 6))
plt.plot(l)
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Analysis of (Loss vs epochs)')
plt.legend()
plt.grid(True)
```

## Prediction

### test data

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python

predicted_probabilities = model.predict(x_test_std)

y_pred = np.argmax(predicted_probabilities, axis=1)

cm = confusion_matrix(y_test, y_pred)

class_names = ['Class 0', 'Class 1', 'Class 2']

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)

fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(cmap=plt.cm.Reds, ax=ax)

plt.title('Neural Network Confusion Matrix')
plt.show()
```

## Model1

```python
model1 = keras.Sequential([
    layers.Dense(units=32, activation='relu', input_shape=(9,)),
    
    layers.Dense(units=16, activation='relu'),
    
    layers.Dense(units=8, activation='relu'),
    
    layers.Dense(units=3, activation='softmax')
])

model1.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

model1.summary()

```

```python
history = model1.fit(x_train_std, y_train, epochs=100, batch_size=10)
```
