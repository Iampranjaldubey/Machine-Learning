```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
from sklearn.metrics import accuracy_score
```

```python

dataset = pd.read_csv('Purchase_Logistic - Purchase_Logistic.csv')

X = dataset.iloc[:, [2, 3]].values 
y = dataset.iloc[:, 4].values
```

### A company is planning to launch a campain for their new car brand  and good like to analyze which customers are most like to purchase the car , So that ads specifically can target them, They consult social network company , that company posses the data from another similar ompany and its now desired to construct a model to acheive this goal

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=11)
```

```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
classifier = LogisticRegression()
classifier.fit(X_train, y_train)
```

## model

```python
classifier.intercept_
```

```python
classifier.coef_
```

## prediction

```python
y_pred = classifier.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

### confusion matrix

```python
# creating the confusion matric
cm = confusion_matrix(y_test, y_pred)
```

```python
cm
```

## Accuracy of model

```python
accuracy_score(y_true=y_test,y_pred=y_pred)
```

## ConfusionMatrixDisplay

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Not Buy', 'Buy'])
disp.plot(cmap=plt.cm.Reds)
plt.title('Confusion Matrix: Car Ad Targeting')
plt.show()
```

```python
def plot_decision_boundary(X, y, model):
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    xx1, xx2 = np.meshgrid(
        np.arange(x1_min, x1_max, 0.01),
        np.arange(x2_min, x2_max, 0.01)
    )

    Z = model.predict(np.c_[xx1.ravel(), xx2.ravel()])
    Z = Z.reshape(xx1.shape)

    plt.contourf(xx1, xx2, Z, alpha=0.3)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor='k')
    plt.xlabel("Age (scaled)")
    plt.ylabel("Salary (scaled)")
    plt.title("Decision Boundary")
    plt.show()
```

```python
plot_decision_boundary(X_train, y_train, classifier)
```

```python
plot_decision_boundary(X_test, y_test, classifier)
```
