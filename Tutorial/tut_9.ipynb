To understand 

```python
def get_adjusted_r2(y_true, y_pred, X):

    r2 = r2_score(y_true, y_pred)
    
    n = X.shape[0] 
    k = X.shape[1] 
    
    adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - k - 1))
    
    return adjusted_r2
```

```python
get_adjusted_r2(y_test,y_predict,x_test)
```

```python
plt.scatter(y_test,y_predict,color = 'red')
```
