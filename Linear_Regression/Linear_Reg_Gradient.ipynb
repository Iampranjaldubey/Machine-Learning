# implementing gradient decent 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cgpa_package.csv")
# print(df.head())

x = df["CGPA"].values
y = df["Package_LPA"].values

n = len(x)
```

# feature scaling

```python
y= (y - np.mean(y)) / np.std(y)
```

# Initialize Parameters

```python
w = 0
b = 0

learning_rate = 0.01
epochs = 1000
```

# Define Cost Function

```python
def compute_cost(w, b):
    y_pred = w * x + b
    return np.mean((y - y_pred) ** 2)
```

# Gradient Descent Loop

```python
cost_history = []

for i in range(epochs):
    
    y_pred = w * x + b
    
    dw = (-2/n) * np.sum(x * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    cost = compute_cost(w, b)
    cost_history.append(cost)
    
print("Final w:", w)
print("Final b:", b)
```

```python
y_pred = w * x + b

plt.scatter(x, y)
plt.plot(x, y_pred, color='red')
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("Gradient Descent Regression")
plt.show()
```

```python
mse =np.mean((y-y_pred)**2)
print(mse)
```

```python
plt.plot(cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost Reduction Over Time")
plt.show()
```
