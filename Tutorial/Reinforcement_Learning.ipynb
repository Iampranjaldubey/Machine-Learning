```python
import random
from typing import List

# environment
class SampleEnvironment:
    def __init__(self):
        self.steps_left = 20
    
    def get_observation(self) -> List[float]:
        return [random.random(), random.random(), random.random()]
       
    def get_action(self) -> List[int]:
        return [0, 1]
    
    def is_done(self) -> bool:
        return self.steps_left == 0
    
    def action(self, action: int) -> float:
        if self.is_done():
            raise Exception("game is over")
        
        self.steps_left -= 1
        
        if action == 1:
            return random.random() + 0.5  # slightly better
        else:
            return random.random()
```

```python
# Agent
class Agent:
    def __init__(self):
        self.total_reward = 0.0
    
    def step(self, env: SampleEnvironment):
        # get current state (not used now, but important conceptually)
        current = env.get_observation()
        
        # get possible actions
        actions = env.get_action()
        
        # choose random action
        action = random.choice(actions)
        print("Chosen action:", action)
        
        # take action and receive reward
        reward = env.action(action)
        print("Reward:", reward)
        
        # accumulate reward
        self.total_reward += reward
```

```python
if __name__ == "__main__":
    env = SampleEnvironment()
    agent = Agent()
    
    while not env.is_done():
        agent.step(env)
    
    print(f"Total reward got: {agent.total_reward:.4f}")
```
