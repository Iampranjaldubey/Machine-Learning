```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
```

```python

dataset = pd.read_csv('Purchase_Logistic - Purchase_Logistic.csv')

X = dataset.iloc[:, [2, 3]].values 
y = dataset.iloc[:, 4].values
```

## Problem

### A company is planning to launch a campain for their new car brand  and good like to analyze which customers are most like to purchase the car , So that ads specifically can target them, They consult social network company , that company posses the data from another similar ompany and its now desired to construct a model to acheive this goal

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=12)
```

```python
classifier = LogisticRegression()
classifier.fit(X_train, y_train)
```

## model

```python
classifier.intercept_
```

```python
classifier.coef_
```

## prediction

```python
y_pred = classifier.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

### confusion matrix

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
cm
```

## Accuracy of model

```python
accuracy_score(y_true=y_test,y_pred=y_pred)
```

## ConfusionMatrixDisplay

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Not Buy', 'Buy'])
disp.plot(cmap=plt.cm.Reds)
plt.title('Confusion Matrix: Car Ad Targeting')
plt.show()
```

### Model Plotting

```python
X_test.shape
```

```python
y = X*classifier.coef_ + classifier.intercept_
```

```python
import numpy as np
import matplotlib.pyplot as plt


exp_y = np.exp(y)
probabilities = exp_y / np.sum(exp_y, axis=1, keepdims=True)

x_plot = X_test.flatten() 
sort_idx = np.argsort(x_plot)

X_sorted = x_plot[sort_idx]
probs_sorted = probabilities[sort_idx]

plt.figure(figsize=(10, 5))
plt.plot(X_sorted, probs_sorted[:, 0], label='Class 0 Prob', color='blue')
plt.plot(X_sorted, probs_sorted[:, 1], label='Class 1 Prob', color='red')
plt.axhline(0.5, color='black', linestyle='--')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
# Make sure your classifier object is trained and available in the environment

def plot_decision_boundary(X_set, y_set, classifier, title):

    age_step = 1
    salary_step = 1000 
    
    # Setup the canvas
    X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = age_step),np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = salary_step))
    
    # Predict over the meshgrid to create the background color map
    Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape)
    
    plt.figure(figsize=(8, 6))
    plt.contourf(X1, X2, Z, alpha = 0.5, cmap = ListedColormap(('red', 'green')))
    plt.xlim(X1.min(), X1.max())
    plt.ylim(X2.min(), X2.max())
    
    for i, j in enumerate(np.unique(y_set)):
        plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                    color = ListedColormap(('red', 'green'))(i), label = f'Purchased: {bool(j)}',
                    edgecolor='white', s=60, alpha=0.9)

    plt.title(title)
    plt.xlabel('Age')
    plt.ylabel('Estimated Salary')
    plt.legend()
    plt.show()

plot_decision_boundary(X_train, y_train, classifier, 'Logistic Regression (Training Set Results)')
```
