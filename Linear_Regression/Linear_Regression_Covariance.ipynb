# Implementing covariance method(linear regression)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("cgpa_package.csv")
df.head()
```

```python
x=df["CGPA"].values
y=df["Package_LPA"].values
```

```python
print(x.shape)
print(y.shape)
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)

print("Mean of x:", x_mean)
print("Mean of y:", y_mean)
```

# Compute Covariance

```python
cov_xy = np.mean((x - x_mean) * (y - y_mean))
print("Covariance:", cov_xy)
```

# Compute Variance of x

```python
var_x=np.mean((x-x_mean)**2)
print(var_x)
```

# Compute Slope (w)

```python
w = cov_xy / var_x
print("Slope:", w)
```

# Compute Intercept (b)

```python
b = y_mean - w * x_mean
print("Intercept (b):", b)
```

```python
def predict(x_new):
    return w*x_new+b
```

```python
y_pred = predict(x)

plt.scatter(x, y, color="blue") 
plt.plot(x, y_pred, color="red")
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("Linear Regression (Covariance Method)")
plt.show()
```

```python
mse = np.mean((y - y_pred)**2)
print("MSE:", mse)
```
