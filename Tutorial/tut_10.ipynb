The objectve of this experiment to understand the classification using tree algorithim.
The aim is to develop and understand how decision tree classifier can be used to categorize data into multiple classes. The experiments focus on undersatnding how tree structure is formed. How spilliting of data occurs using criteria such gni index and how model training and prediction are form. Its also aims to understand how to visualize the decision tree , interpret x structure and evaluate the model  using confusuion matrix and accuracy score.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import accuracy_score
```

```python
from sklearn.tree import plot_tree
from sklearn import tree
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
irisset = datasets.load_iris()
```

```python
irisset.feature_names
```

```python
X = irisset.data
Y = irisset.target
```

```python
X
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state= 12)
```

```python
cf = DecisionTreeClassifier(
    criterion='entropy',
    # random_state=5,
    max_depth=6
)

cf.fit(x_train,y_train)
```

```python
y_pred = cf.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
plt.figure(figsize=(10,6))

plot_tree(
    cf,
    feature_names=["sepal_length","sepal_width","petal_length","petal_width"],
    class_names=['setosa', "versicolor", "virginia"],
    filled= True,
    rounded=True
)

plt.show()
```

```python
text_representation = tree.export_text(
    cf,
     feature_names=["sepal_length","sepal_width","petal_length","petal_width"]
    
)
print(text_representation)
```

```python
df = pd.read_csv("geopolitical_dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
x1_train, x1_test, y1_train, y1_test = train_test_split(x,y, test_size=0.2, random_state= 12)
```

```python
cf1 = DecisionTreeClassifier(
    # criterion='entropy',
    random_state=5,
    max_depth=3
)

cf1.fit(x1_train,y1_train)
```

```python
y1_pred = cf1.predict(x1_test)
```

```python
accuracy_score(y1_test,y1_pred)
```

```python
df.info()
```

```python
plt.figure(figsize=(10,6))

plot_tree(
    cf1,
    feature_names=["military_power_ratio","troop_movement", "nuclear_capability", "nuclear_warheads", "gdp_ratio","sanctions_intensity","trade_dependency", "ideology_distance","political_stability", "historical_conflict","border_distance", "news_sentiment", "nationalism_index", "leader_aggression"],
    # class_names=["War","No war"],
    filled= True,
    rounded=True
)

plt.show()
```

```python
rf = RandomForestClassifier (
    n_estimators= 100,
    criterion= 'gini',
    max_depth= 3,
    random_state= 42
)
rf.fit(x1_train, y1_train)
```

```python
rf_predict = rf.predict(x1_test)
```

```python
accuracy_score(y1_test,rf_predict)
```

```python
plt.figure(figsize=(15,8)) 

first_tree = rf.estimators_[0]

plot_tree(
    first_tree,
    feature_names=["military_power_ratio","troop_movement", "nuclear_capability", "nuclear_warheads", "gdp_ratio","sanctions_intensity","trade_dependency", "ideology_distance","political_stability", "historical_conflict","border_distance", "news_sentiment", "nationalism_index", "leader_aggression"],
    filled=True,
    rounded=True,
    max_depth=2, 
    fontsize=10
)

plt.show()
```
