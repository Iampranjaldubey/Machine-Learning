```python
from keras.models import Sequential
from keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
from sklearn.metrics import mean_absolute_error , mean_squared_error, r2_score
import pandas as pd
```

```python
df = pd.read_csv("BostonHousing.csv")
```

```python
X = df.iloc[:,0:11]
y = df.iloc[:,13]
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size= 0.75, random_state= 11)
```

```python
scaler = StandardScaler()

x_train_std = scaler.fit_transform(x_train)
```

```python
model = Sequential()
```

```python
model.add(Dense(9,input_dim = 11,activation='relu'))
```

```python
model.add(Dense(30,input_dim = "11",activation='relu'))
```

```python
model.add(Dense(1))
```

```python
model.compile(loss = 'mean_squared_error',optimizer = 'adam' , metrics= ['MSE'])
```

```python
history = model.fit(x_train_std,y_train,epochs=100,batch_size=100)
```

```python
y_pred = model.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

## Plot

```python
l = history.history['loss']
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 6))
plt.plot(l)
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Analysis of (Loss vs epochs)')
plt.legend()
plt.grid(True)
```
