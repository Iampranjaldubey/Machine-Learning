```python
import numpy as np
import pandas as pd
```

```python
from sklearn.cluster import KMeans
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("logistics_2000_dataset.csv")
```

```python
df.describe()
```

```python
x = df[['latitude','longitude']]
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
x.shape
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
interia = []
k_range = range(1,15)

for k in k_range:
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(x_scaled)
    interia.append(kmeans.inertia_)
    
```

```python
plt.plot(k_range, interia, marker = 'o')
plt.xlabel('K')
plt.ylabel('Interia')
plt.title('Elbow method')
plt.grid(True)
plt.show()
```

```python
interia_4 = []
kmeans4 = KMeans(n_clusters=4,random_state=42)
kmeans4.fit(x_scaled)
interia_4.append(kmeans4.inertia_)
```

```python
interia_4
```

```python
plt.plot(1, interia_4, marker = 'o')
plt.xlabel('K')
plt.ylabel('Interia')
plt.title('Elbow method')
plt.grid(True)
plt.show()
```

```python
interia4range = []
k_range = range(1,15)

for k in k_range:
    kmeans4range = KMeans(n_clusters=4,random_state=42)
    kmeans4range.fit(x_scaled)
    interia4range.append(kmeans4range.inertia_)
    
```

```python
plt.plot(k_range, interia4range, marker = 'o')
plt.xlabel('K')
plt.ylabel('Interia')
plt.title('Elbow method')
plt.grid(True)
plt.show()
```

```python
centre = kmeans4range.cluster_centers_
```

```python
centre
```

```python
df['cluster'] = kmeans4range.fit_predict(x_scaled)
```

```python
centre_original = scaler.inverse_transform(centre)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['latitude'],df['longitude'], c = df['cluster'], s=10)
plt.scatter(centre_original[:, 0], centre_original[:, 1], s=300, marker='X', color='red')


plt.xlabel('latitude')
plt.ylabel('longitude')

plt.show()
```

### Distance based clustering

```python
x_distance = df['distance_km'].values.reshape(-1,1)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_distance)
```

```python
kmeans = KMeans(n_clusters=8,random_state=42)
kmeans.fit(x_scaled)
centre = kmeans.cluster_centers_
```

```python
df['cluster'] = kmeans4range.fit_predict(x_scaled)
```

### accessibility

```python
x_distance = df['accessibility'].values.reshape(-1,1)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_distance)

kmeans = KMeans(n_clusters=8,random_state=42)
kmeans.fit(x_scaled)
centre = kmeans.cluster_centers_

df['cluster'] = kmeans4range.fit_predict(x_scaled)

centre_original = scaler.inverse_transform(centre)

plt.figure(figsize=(8,6))

plt.scatter(df['latitude'],df['longitude'], c = df['cluster'], s=10)
plt.scatter(centre_original[:, 0], centre_original[:, 1], s=300, marker='X', color='red')


plt.xlabel('latitude')
plt.ylabel('longitude')

plt.show()
```
